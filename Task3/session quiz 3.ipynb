```python
from numpy import pi

def cylinder(r, h):
    A = 2 * pi * r**2 + 2 * pi * r * h
    V = pi * r**2 * h
    return A, V
result = cylinder(5, 7)
print(cylinder(5, 7))

print("ratio =", result[0]/result[1])
```

```python
import numpy as np
from functools import reduce

def stdev(p, N):
    y = p
    mu = 0
    sigma = 0
    for i in range(0,N):
        mu +=  y[i]
    mu = mu/N
    for i in range(0,N): 
        sigma += 1/N * (y[i] - mu)**2
    return sigma**0.5

array = np.random.normal(0, 5, 1000)
print(stdev(array, len(array)))

mu = 0
sigma = 0
y = array
N = len(array)

print(reduce(lambda x, y: x + y, [(1/N * (y[i] - reduce(lambda x, y: x + y, [1/N * y[i] for i in range(N)]))**2) for i in range(N)])**0.5)
print(np.std(array))
```
