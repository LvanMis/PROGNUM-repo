```python
masses = [1.9891e+30, 1.8986e+27, 5.6846e+26, 
          1.0243e+26, 8.6810e+25, 5.9736e+24, 
          4.8685e+24, 6.4185e+23, 3.3022e+23, 
          7.349e+22, 1.25e22]
radii  = [696340, 71492, 60268, 24764, 
          25559, 6378.1, 6051.8, 3396.2, 
          2439.7, 1738.1, 1195]
names  = ['Sun', 'Jupiter', 'Saturn', 'Neptune', 
          'Uranus', 'Earth', 'Venus', 'Mars', 
          'Mercury', 'Moon', 'Pluto']

G = 6.67*10**-11

v_esc1 = []

for n in range(0, len(masses)):
    r = radii[n]
    M = masses[n]
    v_esc = ((2*G*M)/r)**0.5
    v_esc1.append(v_esc)
    print(f"For {names[n]} with  mass {masses[n]} the escape velocity is {v_esc1[n]}")
    
    dictionary = dict(zip(names, v_esc1))
```

```python
print(dictionary)
```
