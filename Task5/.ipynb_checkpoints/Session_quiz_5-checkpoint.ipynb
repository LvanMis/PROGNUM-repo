```python
# session quiz 5.6

import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
ab = np.concatenate((a, b))
print(ab)
print(f'the shape is {ab.shape}')

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
# the difference is that the second group is 2dimensional
ab = np.concatenate((a, b))
print(ab)
print(f'the shape is {ab.shape}')

ab = np.concatenate((a, b), axis = 0)
print(ab)
print(f'the shape is {ab.shape}')

ab = np.concatenate((a, b), axis = 1)
print(ab)
print(f'the shape is {ab.shape}')
# the difference is the way that the matrices are stuck together. O seems to be the default, which stacks them vertically. 1 seems to stack them horizontally
```

```python
# session quiz 5.12

from scipy.constants import c

class Galaxy:    
    """Class for redshift calculation of galaxies"""  
        
    def __init__(self, name, calciumHline=4000):
        self.name = name
        self.caH = calciumHline
    def redshift(self):        
        caH_0 = 3968.5      # Calcium H line in rest frame
        z = (self.caH - caH_0)/self.caH
        return z
    def velocity(self, c, z):
        vel_galaxy = (z**2 *c - c) / (1 + z**2)
        return vel_galaxy
    
red = Galaxy('name', calciumHline=4000).redshift()
Galaxy('name', calciumHline=4000).velocity(c, red)
```

```python
# session quiz 5.13

import numpy as np
from matplotlib.pyplot import figure, show, imshow
from matplotlib.patches import Ellipse
Nx = 400
Ny = 300
data = np.random.random((Ny,Nx))


fig = figure()
frame = fig.add_subplot(1,2,1, aspect=1)
semi_minor = 50
semi_major = 100
pos_angle = np.degrees(30)
frame.imshow(data, interpolation='none', origin="lower", cmap='gray')

width = 2 * semi_minor
height = 2 * semi_major

ellipse = Ellipse((150, 200), width, height, angle=pos_angle, fill = False, edgecolor = 'red')
frame.set_xlim(0,Nx)    # Limits are not automatically adjusted when patches are used
frame.set_ylim(0,Ny)
frame.add_patch(ellipse)
```
