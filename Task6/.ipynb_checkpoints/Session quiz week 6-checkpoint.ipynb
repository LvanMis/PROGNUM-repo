```python
# session quiz 6.1

import numpy as np
import astropy.io.fits as fits
import imageio.v2 as io
from scipy import stats
from matplotlib import pyplot as plt
from mpl_toolkits.axes_grid1 import make_axes_locatable

#from matplotlib.axes.Axes import axvline
dataa = fits.open('m101.fits')

dataa.info()
# there are 2 HDU's
# the BinTableHDU is where the image info is stored
# the number types are integers

head = dataa[0].header
for key in head[:5]:
    print(f"{key}= {head[key]}")
# BITPIX = 16

dat = dataa[0].data
print('shape is', np.shape(dat))

# So the shape does correspond

print('dtype =',dat.dtype)
# according to Numpy: describes how the bytes in the fixed-size block of memory corresponding to an array item should be interpreted
# The specific dtype I get means 'big-endian 2-byte integer'
# Python says: 'Signed and unsigned 16-Bit (2 Byte) integer types' for both of these so they correspond

flatdat = dat.flatten()
s = stats.describe(flatdat)
print(s)
print(stats.mode(dat.flat))
print('median=',np.median(dat.flat))
fig, ax = plt.subplots(1, 1, figsize = (5, 5))

ax.hist(flatdat)
ax.axvline(np.float64(4795.924268422927), color = 'k', label = 'mean')
ax.axvline(4102.0, color='k', linestyle = '--', label = 'median')
ax.axvline(np.int16(3551), color='k', linestyle = ':', label = 'mode')
ax.legend()

fig, ax = plt.subplots(1, 1, figsize = (5, 5))
dat2 = np.asarray(dat)
med = np.median(dat2)
im = ax.imshow(dat2, origin = 'lower', vmin = med)
ax.set_xlabel(head['CTYPE1'])
ax.set_ylabel(head['CTYPE2'])
divider = make_axes_locatable(ax)
colbarframe = divider.append_axes("right", size="5%", pad=0.15)

fig.colorbar(im, cax=colbarframe)
plt.show()

head['BUNIT'] = 'Counts'
logdat = np.log10(dat)
print("Min. and Max. values of log10(dat):", logdat.min(), logdat.max())

headnew = fits.PrimaryHDU(logdat)
headnew.header = head
headnew.writeto('m102.fits', overwrite=True, output_verify='ignore')
# to ignore error messages, use: output_verify='ignore'
```

```python
# session quiz 6.3

import numpy as np
import matplotlib.pyplot as plt

dataa = fits.open('m101.fits')
dat = dataa[0].data
dat2 = np.asarray(dat)[:20, :20]
plt.imshow(dat2, origin = 'lower')
plt.show()

dat3 = dat2.flatten()
total = 0
for i in dat3:
    total += i**2
x_rms = np.sqrt(1 / len(dat3) * total)
print('x_RMS =', x_rms)
print('mean =', np.mean(dat3))
print('std =', np.std(dat3))
# std has as the numerator in the root as (xi - mean)**2. For the x_RMS it's just xi**2.
# x_RMS, because we don't necessarily care about how much it deviates. If we want to filter out the noise we simply 
# need to know which values it's between
```

```python
# session quiz 6.5

data0 = fits.open('noisemap0.fits')
data1 = fits.open('noisemap1.fits')
data2 = fits.open('noisemap2.fits')

data0 = data0[0].data
data1 = data1[0].data
data2 = data2[0].data

print('std0 =', np.std(data0))
print('std1 =', np.std(data1))
print('std2 =', np.std(data2))

data = np.stack((data0, data1, data2), axis = 0)
print(data.shape)
mean = np.mean(data, axis = 0)
print('3D std =', np.std(mean))

count = 0
total = 0
for i in data:
    count += 1
    total += np.std(i)**2
sigma = 1/count * np.sqrt(total)
print('Regular std =', sigma)
```

```python
# session quiz 6.10
from astroquery.gaia import Gaia
import sqlite3 as sqlite
import pandas as pd
from astropy.table import Table

query = """
SELECT TOP 50
    source_id,
    ra,
    dec,
    parallax,
    parallax_error,
    phot_g_mean_mag,
    phot_bp_mean_mag,
    phot_rp_mean_mag
FROM gaiadr3.gaia_source
WHERE
    parallax / parallax_error > 5
    AND phot_g_mean_mag IS NOT NULL
    AND phot_bp_mean_mag IS NOT NULL
    AND phot_rp_mean_mag IS NOT NULL
"""


filtered = Gaia.launch_job(query)
results = filtered.get_results()
df = results.to_pandas()

table = Table.from_pandas(df)
table.write('astroquery_gaia.fits', format='fits', overwrite=True)
```

```python
# session quiz 6.6

from astropy.io import fits
import numpy as np
from astropy.table import Table
import pandas as pd

hdul = fits.open('APOGEE_stars.fits')
data_fits = hdul[1].data  # Often, the first extension contains the table
colnames = hdul[1].columns.names # List of the column names
data = hdul[1].data
data_np = np.array(data_fits).byteswap().newbyteorder()
data = pd.DataFrame(data_np)
# I needed to convert it a bunch of times because it kept saying that the endian types were wrong

data['DATAFLAG'] = ((data['STARFLAG'] == 0) &
    (data['ASPCAPFLAG'] == 0) &
    (data['SNR'] > 10) &
    (data['GAIAEDR3_PARALLAX'] / data['GAIAEDR3_PARALLAX_ERROR'] > 5) &
    ~np.isnan(data['GAIAEDR3_PHOT_G_MEAN_MAG']) &
    ~np.isnan(data['GAIAEDR3_PHOT_BP_MEAN_MAG']) &
    ~np.isnan(data['GAIAEDR3_PHOT_RP_MEAN_MAG'])).astype(int)

filtered_data = data[data['DATAFLAG'] == 1]
# print(filtered_data)

data['ABS_MAG'] = np.where(data['DATAFLAG'] == 1, (
    data['GAIAEDR3_PHOT_G_MEAN_MAG'] +
    5 * np.log10(data['GAIAEDR3_PARALLAX'] / 1000) +
    5), -1.0)

print(data)

table = Table.from_pandas(data)
table.write('APOGEE_update.fits', format='fits', overwrite=True)
```
